# PANDAS
  - Pandas are used to analyze the data.

##### DEF:-- Pandas is a open-source library used to analyze data , manipulate the data and cleaning the data.
             - It provides the data structures like Series and Dataframes for handling the  structured data easily.
             - Pandas library built using the numpy, which is heavily used to data Manipulation, Visualisation and building machine learning models.

##### There are two main data structures in the pandas, they are SERIES and DATA FRAMES. 
      - The default way to store the data is data frames.
      - manipulating the data quickly is one of the most important skill in data analysis
     
             

#### 1.SERIES
     - Series is a Data Structure, which has same datatypes(numeric, charecter, datetime etc..)
     - Series is like a one-dimensional array which will have column like values like excel and csv sheets.
     - ex:- Its like rows and columns in the excel sheet.
             data = [10, 20, 30, 40]
             syntax:--   pd.Series(data, index=[values])

#### 2.DATA FRAMES
     - Dataframes(DF) is a data structure which has different datatypes.
     - DF is like a two dimensional array,which has rows and columns.
     - ex:- rows and columns in the excel sheet.
          data = {'Name': ['Alice', 'Bob', 'Charlie'],
            'Age': [25, 30, 35],
            'Salary': [50000, 60000, 70000]} 
            - SYNTAX:--
                 pd.DataFrame(data, index=[values])

## Selecting the subset of the dataframes
  - If we select the third row or any row, it will automaticlly gives error.
  - Even after changing the index as to any number we cannot get the particular element directly from the data.
  - To get the sub-elements, we have a concept called positional arguments:-- loc and iloc.

### Position or Label based indexing:-
   - TWO Main ways of indexing df:-
     1. Position based indexing by using pd.iloc()
     2. Label based indexing by using pd.loc()
   - By using both methods, we can do the below operations:-
     - Selecting single elements/cells.
     - Selecting single and multiple rows.
     - Selecting single and multiple columns.
     - Selecting multiple rows and columns. 

```python
df1 = pd.read_csv("C:/Users/rakes/Downloads/market_fact.csv")
df1
```

#### Subsetting the rows based on the conditions

```python
## selecting the sales 2000>sales<3000
import pandas as pd
df=pd.read_csv(r"C:\Users\rakes\Downloads\market_fact.csv")
df
```

```python
# Selecting sales > 3000/-
df[df.Sales>3000]
```

```python
# Selecting sales, 2000>sales<3000 , profit>100
df[(df.Sales>2000) & (df.Sales<3000) & (df.Profit > 100)]
```

```python
# 2000<sales,or profit >100
df[(df.Sales<2000) | (df.Profit>100)]
```

```python
# Selecting the required columns and rows.
## 2000>Sales<3000
## selecting the sales,discount, order_quantity, and profit columns.
df.loc[(df.Sales>2000)&(df.Sales<3000)&(df.Profit>100),['Sales','Discount','Order_Quantity','Profit']]

```

```python
## Using == and !==
df[df.Profit == 636.18]
```

```python
df[df.Profit != 636.18]
```

```python
## For instance, a businessman has given some customer id's from a particular region, so based on the cust_id we need to find out the customer data.
## By using .isin() function, we can find it out easily.

customer_ids = ['Cust_1818','Cust_1749','Cust_45','Cust_1266','Cust_1006','Cust_498']
df[df['Cust_id'].isin(customer_ids)]
```
